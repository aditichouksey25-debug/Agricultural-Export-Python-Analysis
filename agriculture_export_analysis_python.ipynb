```python
rice=df[df["commodity"]=="Rice"]
print(rice)
```

```python
rice=rice[rice["partner"]!="World"]
print(rice)
```

```python
soyabean=df[df["commodity"]=="Soyabean"]
print(soyabean)
```

```python
soyabean=soyabean[soyabean["partner"]!="World"]
print(soyabean)
```

```python
rice.groupby("partner")["trade_value"].sum().sort_values(ascending=False).head(10)
```

```python
soyabean.groupby("partner")["trade_value"].sum().sort_values(ascending=False).head(10)
```

```python
top10_rice= (rice.groupby("partner")["trade_value"].sum().sort_values(ascending=False).head(10))
print(top10_rice)
```

```python
top10_soyabean=(soyabean.groupby("partner")["trade_value"].sum().sort_values(ascending=False).head(10))
print(top10_soyabean)
```

```python
import matplotlib.pyplot as plt
top10_rice.plot(kind="bar",figsize=(8,5),rot=45,color="forestgreen")
plt.title("Top 10 Rice Export Markets",fontsize=15)
plt.xlabel("partner",fontsize=12)
plt.ylabel("trade_value(Billion USD)",fontsize=12)
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.savefig("TOP 10 RICE MARKETS.PNG",bbox_inches="tight",dpi=300)
plt.show()
```

```python
top10_soyabean.plot(kind="bar",color="orange",rot=45,figsize=(8,5))
plt.title("Top 10 Soyabean Markets",fontsize=15)
plt.xlabel("partner",fontsize=12)
plt.ylabel("trade_value(Billion USD)",fontsize=12)
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.savefig("TOP 10 SOYABEAN MARKETS")
plt.show()
```

```python
df["value_per_kg"]=df["trade_value"]/df["net_weight"]
```

```python
rice=df[df["commodity"]=="Rice"]
rice["value_per_kg"].head()
```

```python
rice["value_per_kg"].mean()
```

```python
soyabean= df[df["commodity"]=="Soyabean"]
soyabean["value_per_kg"].head()
```

```python
soyabean["value_per_kg"].mean()
```

```python
round(rice["value_per_kg"].mean(),2)
```

```python
round(soyabean["value_per_kg"].mean(),2)
```

```python
rice_avg=round(rice["value_per_kg"].mean(),2)
```

```python
soyabean_avg=round(soyabean["value_per_kg"].mean(),2)
```

```python
comparison=pd.Series({"rice":rice_avg,"soyabean":soyabean_avg})
comparison.plot(kind="bar",figsize=(8,5),rot=45,color="purple")
plt.title("Average Export Value Per Kg")
plt.xlabel("commodity")
plt.ylabel("value per kg")
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.savefig("rice vs soyabean value per kg",dpi=300,bbox_inches="tight")
plt.show()
```

```python
df["year"].unique()
```

```python
rice_commodity=(df[df["commodity"]=="Rice"].groupby(["year"])["trade_value"].sum())
rice_commodity.plot(figsize=(8,5),marker="o")
plt.title("year-wise export trend of rice")
plt.xlabel("year")
plt.ylabel("trade value")
plt.grid(True,linestyle="--",alpha=0.7)
plt.savefig("rice_yearly_trend.png",dpi=300,bbox_inches="tight")
plt.show()
```

```python
soyabean_commodity=(df[df["commodity"]=="Soyabean"].groupby(["year"])["trade_value"].sum())
rice_commodity.plot(figsize=(8,5),marker="o")
plt.title("year-wise export trend of soyabean")
plt.xlabel("year")
plt.ylabel("trade value")
plt.grid(True,linestyle="--",alpha=0.7)
plt.savefig("soyabean_yearly_trend.png",dpi=300,bbox_inches="tight")
plt.show()
```
